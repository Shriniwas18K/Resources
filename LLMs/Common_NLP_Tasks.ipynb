### Text Generation: These predict next tokens with scores. ChatGPT is built on these.
###### This is basically sentence completion hence OpenAI introduced ChatCompletions API earlier.
###### Each model has its own prompting format

```python
generator = pipeline(
    "text-generation", # 957 MB
    model = "Qwen/Qwen2.5-0.5B-Instruct"
)
```

```python
request = "Tell me about ancient philopshers of India."

prompt = f"""
  User: {request}.
  Assistant:
"""

NUMBER_OF_RESPONSES = 2

result = generator(
    prompt,
    max_new_tokens = 200,
    num_return_sequences = NUMBER_OF_RESPONSES
)

for r in result:
  print(r["generated_text"])
```

## Text classification:

### Multi genre NLI(Natural Language Inferencing)
Identify logical relationship (entails,contradicts,neutral) b/w premise and hyopthesis

```python
mnli = pipeline(
    "text-classification", # 1363mb
    model = "FacebookAI/roberta-large-mnli"
)
```

```python
premise = "Indian cricket team didnt win 2025 world cup"
hypothesis = "Indian cricket teams plays the best cricket."
mnli(
    f"{premise} </s> {hypothesis}"
)
```

### QNLI
Identify wheter question can be answered from given context

```python
qnl = pipeline(
    "text-classification", # 418 mb
    model = "cross-encoder/qnli-electra-base"
)
```

```python
qs = "Who won world cup 2025?"
context = "The final match of world cup 2025 was played between India and Australia."
qnl(
    f"{qs} </s> {context}",
    top_k=None
) # label_0 means not entails, means qs cannot be answered from context
```

### QQP: Quora Question Pairs
identifies wheter two questions are meaning same

```python
qqp = pipeline(
    "text-classification", # 1639mb
    model = "howey/roberta-large-qqp"
)
```

```python
q1 = "Whos captain of Indian cricket team?"
q2 = "Who leads Indian Cricket?"
qqp(f"{q1} [SEP] {q2}") # LABEL_1 means qs are similar, score indicates to what extent, near 1 means highly similar
```

## Chatbot

```python
import torch
from transformers import AutoModelForCausalLM,AutoTokenizer
MODEL_NAME = "Qwen/Qwen2.5-0.5B-instruct"
MAX_CHAT_HISTORY = 5
MAX_NEW_TOKENS = 200
device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME
)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    dtype="auto",
    device_map="auto"
).to(device)
```

```python
chat_history = []

def generate_response(prompt):
  global chat_history
  messages = [
      {
          "role" : "system",
          "content" : "You are medival era scholar in mathematics."
      }
  ]
  for message in chat_history[-MAX_CHAT_HISTORY:]:
    role,content = message.split(": ",1)
    messages.append({
        "role" : role,
        "content" : content
    })
  messages.append({
      "role":"user",
      "content":prompt
  })

  text = tokenizer.apply_chat_template(
      messages,
      tokenize = False,
      add_generation_prompt = True
  )

  model_input = tokenizer([text],return_tensors = "pt").to(device)

  outputs = model.generate(
      **model_input,
      max_new_tokens = MAX_NEW_TOKENS,
      pad_token_id = tokenizer.eos_token_id
  )

  generated_ids = outputs[:,model_input.input_ids.shape[-1]:]

  response = tokenizer.batch_decode(
      generated_ids,
      skip_special_tokens = True
  )[0]

  chat_history.append(f"user: {prompt}")
  chat_history.append(f"assistant: {response}")

  print("assistant: ",response)
```

```python
while True:
  prompt = input("user: ")
  generate_response(prompt)
```
